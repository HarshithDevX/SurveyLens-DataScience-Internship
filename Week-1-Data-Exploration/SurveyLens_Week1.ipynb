## 6. Import Required Python Libraries

```python
from IPython.display import display
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("SurveyLens Week 1 Project Started!")
```

## 7. Load and Preview the Dataset

Load the public UCI Adult training CSV. The `?` marker is interpreted as a missing value.

```python
# SurveyLens - Week 1
# Step 2: Load the Public Survey Dataset

import pandas as pd

# Official UCI dataset URL
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.data"

# Dataset column names
columns = [
    "age", "workclass", "fnlwgt", "education",
    "education_num", "marital_status", "occupation",
    "relationship", "race", "sex", "capital_gain",
    "capital_loss", "hours_per_week",
    "native_country", "income"
]

# Load the dataset
df = pd.read_csv(
    url,
    names=columns,
    na_values="?",
    skipinitialspace=True
)

print("Dataset Loaded Successfully!")

print("\nTotal Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

# Display first 5 records
display(df.head())
```

## 8. Explore Dataset Structure and Numeric Statistics

```python
# SurveyLens - Week 1
# Step 2: Understanding the Dataset

print("========== DATASET INFORMATION ==========")
df.info()

print("\n========== BASIC STATISTICS ==========")
display(df.describe())

print("\n========== ALL COLUMNS ==========")
print(df.columns.tolist())

print("\n========== DATASET SIZE ==========")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

## 9. Explore Four Categorical Variables

Counts and percentages use non-missing answers for each category.

```python
# SurveyLens - Week 1
# Step 3: Exploring Categorical Columns

categories = [
    "education",
    "workclass",
    "occupation",
    "marital_status"
]

for column in categories:
    print(f"\n========== {column.upper()} ==========")

    # Count responses in each category
    print(df[column].value_counts())

    # Calculate percentage distribution
    print("\nPercentage Distribution:")
    print(df[column].value_counts(normalize=True).mul(100).round(2))
```

## 10. Identify Missing Values

We **identify**, but do not remove or replace, missing records in Week 1. Handling them belongs to the cleaning stage.

```python
# SurveyLens - Week 1
# Step 4: Missing Value Analysis

print("========== MISSING VALUES ==========")

# Count missing values in each column
missing_values = df.isnull().sum()

print(missing_values)

print("\n========== MISSING PERCENTAGE ==========")

# Calculate percentage of missing values
missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage.round(2))

print("\n========== COLUMNS NEEDING CLEANING ==========")

# Display columns containing missing values
print(missing_values[missing_values > 0])
```

## 11. Compare Working Hours Across Education Groups

These are descriptive group averages and do not establish that education causes different working hours.

```python
# SurveyLens - Week 1
# Step 5: Grouped Statistical Analysis

print("========== AVERAGE WORKING HOURS BY EDUCATION ==========")

# Group data by education and calculate average working hours
education_hours = df.groupby("education")["hours_per_week"].mean()

# Sort results from highest to lowest
education_hours = education_hours.sort_values(ascending=False)

# Display results
print(education_hours.round(2))

print("\n========== TOP 5 EDUCATION GROUPS ==========")
print(education_hours.head(5).round(2))
```

## 12. Visualization 1 — Top 10 Occupations

Each bar represents the number of records with that occupation. Missing occupations are excluded from the ranking.

```python
# SurveyLens - Week 1
# Step 6: Top 10 Occupations Bar Chart

import matplotlib.pyplot as plt
import seaborn as sns

# Get the top 10 occupation categories
top_occupations = df["occupation"].value_counts().head(10)

# Create chart
plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_occupations.values,
    y=top_occupations.index,
    color="steelblue"
)

# Add title and axis labels
plt.title("Top 10 Occupations in the UCI Adult Dataset",
          fontsize=15, fontweight="bold")

plt.xlabel("Number of Records", fontsize=12)
plt.ylabel("Occupation Category", fontsize=12)

# Display values on the bars
for i, value in enumerate(top_occupations.values):
    plt.text(value + 20, i, str(value), va="center")

plt.xlim(0, top_occupations.max() * 1.15)

plt.tight_layout()

# Save chart for LinkedIn and GitHub
plt.savefig("top_10_occupations.png", dpi=300, bbox_inches="tight")

plt.show()
```

## 13. Visualization 2 — Missing Data Quality

This chart highlights the columns we may need to clean in Week 2. Percentages are based on all **32,561 records**.

```python
# Additional visualization 2: Missing values by column
missing_counts = df.isnull().sum()
missing_counts = missing_counts[missing_counts > 0].sort_values()

plt.figure(figsize=(9, 4.8))
bars = plt.barh(missing_counts.index, missing_counts.values, color="#2979A5")
plt.title("Missing Values by Column — UCI Adult Dataset", fontsize=14, fontweight="bold")
plt.xlabel("Number of Missing Records")
plt.ylabel("Column")
plt.xlim(0, missing_counts.max() * 1.18)
for bar in bars:
    value = int(bar.get_width())
    plt.text(value + missing_counts.max() * 0.02,
             bar.get_y() + bar.get_height() / 2,
             f"{value:,} ({value / len(df):.2%})", va="center")
plt.tight_layout()
plt.savefig("missing_values_by_column.png", dpi=300, bbox_inches="tight")
plt.show()
```

## 14. Visualization 3 — Working Hours by Education

The labels include the number of records per education group, which is important when comparing averages.

```python
# Additional visualization 3: Compare average working hours across education groups
# Include all education groups, ranked by the average weekly hours.
hours_by_education = df.groupby("education")["hours_per_week"].agg(["count", "mean"])
hours_by_education = hours_by_education.sort_values("mean", ascending=True)

plt.figure(figsize=(10, 7.5))
plt.barh(hours_by_education.index, hours_by_education["mean"], color="#397D73")
plt.title("Average Weekly Working Hours by Education — UCI Adult", fontsize=14, fontweight="bold")
plt.xlabel("Mean Hours Worked per Week")
plt.ylabel("Education Level")
plt.xlim(0, hours_by_education["mean"].max() * 1.18)
for i, (mean_value, n) in enumerate(zip(hours_by_education["mean"], hours_by_education["count"])):
    plt.text(mean_value + 0.6, i, f"{mean_value:.1f} h  (n={n:,})", va="center", fontsize=8.5)
plt.tight_layout()
plt.savefig("working_hours_by_education.png", dpi=300, bbox_inches="tight")
plt.show()
```

## 15. Optional Stretch Goal — Cross-tabulation and Heatmap

The table uses `pd.crosstab`. Each row sums to approximately 100%, so the heatmap compares **income composition within each education group** rather than raw group size. These patterns are **associations, not causal effects**.

```python
# SkillMe optional stretch goal: cross-tabulation + heatmap
# Percentages are calculated WITHIN each education group so groups of
# different sizes can be compared more fairly.
income_education_pct = pd.crosstab(
    df["education"], df["income"], normalize="index"
).mul(100)

# Arrange education categories in the dataset's ordinal education_num order.
education_order = (
    df.groupby("education")["education_num"]
      .median()
      .sort_values(ascending=False)
      .index
)
income_education_pct = income_education_pct.reindex(education_order)

print("Education vs Income — Row Percentages")
display(income_education_pct.round(1))

plt.figure(figsize=(9.5, 9))
sns.heatmap(income_education_pct, annot=True, fmt=".1f", cmap="YlGnBu",
            linewidths=0.5, cbar_kws={"label": "Percentage of Records (%)"})
plt.title("Income Category by Education (Row %) — Historical UCI Adult Data",
          fontsize=13, fontweight="bold", pad=14)
plt.xlabel("Income Category")
plt.ylabel("Education Level")
plt.tight_layout()
plt.savefig("education_income_heatmap.png", dpi=300, bbox_inches="tight")
plt.show()

# Print a valid data-driven finding instead of guessing a heatmap result.
high_income_column = next((col for col in income_education_pct.columns if col.strip() == ">50K"), None)
if high_income_column is not None:
    top_group = income_education_pct[high_income_column].idxmax()
    top_pct = income_education_pct.loc[top_group, high_income_column]
    print(f"Highest proportion of >50K records: {top_group} ({top_pct:.1f}% within that group)")
```
